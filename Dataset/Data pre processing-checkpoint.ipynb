```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
BASE_PATH = Path(r"C:\Users\moham\Desktop\TechFest Hackathon\NTU TechFest-20260124T023846Z-3-001\NTU TechFest")
```

```python
glassdoor = pd.read_csv(BASE_PATH / "Glassdoor.csv")
indeed = pd.read_csv(BASE_PATH / "Indeed.csv")
prosple = pd.read_csv(BASE_PATH / "Prosple.csv")
jobstreet = pd.read_csv(BASE_PATH / "Jobstreet.csv")
efinancial = pd.read_csv(BASE_PATH / "efinancialcareers.csv")
mycareers = pd.read_csv(BASE_PATH / "Mycareersfuture.csv")
```

```python
for name, df in {
    "Glassdoor": glassdoor,
    "Indeed": indeed,
    "Prosple": prosple,
    "Jobstreet": jobstreet,
    "Efinancialcareers": efinancial,
    "Mycareersfuture": mycareers
}.items():
    print(name, df.shape)
```

```python
def normalize_columns(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace("&", "and", regex=False)
        .str.replace(r"[^\w\s]", "", regex=True)
        .str.replace(r"\s+", "_", regex=True)
    )
    return df
```

```python
glassdoor = normalize_columns(glassdoor)
indeed = normalize_columns(indeed)
prosple = normalize_columns(prosple)
jobstreet = normalize_columns(jobstreet)
efinancial = normalize_columns(efinancial)
mycareers = normalize_columns(mycareers)
```

```python
for name, df in {
    "Glassdoor": glassdoor,
    "Indeed": indeed,
    "Prosple": prosple,
    "Jobstreet": jobstreet,
    "Efinancial": efinancial,
    "MyCareersFuture": mycareers
}.items():
    print(f"\n{name} columns:")
    print(df.columns.tolist())
```

```python
glassdoor_clean = pd.DataFrame({
    "title": glassdoor["title"],
    "company": glassdoor["company"],
    "location": glassdoor["location"],
    "employment_type": np.nan,
    "job_type": np.nan,
    "salary": glassdoor.get("salary"),
    "posted_date": np.nan,
    "description": glassdoor["description"],
    "rating": glassdoor.get("rating"),
    "source": "glassdoor",
    "url": glassdoor["url"]
})
```

```python
indeed_clean = pd.DataFrame({
    "title": indeed["title"],
    "company": indeed["company"],
    "location": indeed["location"],
    "employment_type": indeed.get("job_type"),
    "job_type": indeed.get("job_type"),
    "salary": np.nan,
    "posted_date": np.nan,
    "description": indeed["description"],
    "rating": np.nan,
    "source": "indeed",
    "url": indeed["url"]
})
```

```python
prosple_clean = pd.DataFrame({
    "title": prosple["title"],
    "company": prosple["company"],
    "location": prosple["location"],
    "employment_type": prosple.get("type"),
    "job_type": prosple.get("type"),
    "salary": prosple.get("salary"),
    "posted_date": prosple.get("startdate"),
    "description": prosple["description"],
    "rating": np.nan,
    "source": "prosple",
    "url": prosple["url"]
})
```

```python
jobstreet_clean = pd.DataFrame({
    "title": jobstreet["title"],
    "company": jobstreet["company"],
    "location": jobstreet["location"],
    "employment_type": jobstreet.get("job_type"),
    "job_type": jobstreet.get("job_type"),
    "salary": jobstreet.get("salary"),
    "posted_date": jobstreet.get("posted"),
    "description": jobstreet["description"],
    "rating": jobstreet.get("review"),
    "source": "jobstreet",
    "url": jobstreet["url"]
})
```

```python
efinancial_clean = pd.DataFrame({
    "title": efinancial["title"],
    "company": efinancial["company"],
    "location": efinancial["location"],
    "employment_type": efinancial.get("employment_type"),
    "job_type": efinancial.get("workplace_model"),
    "salary": efinancial.get("salary"),
    "posted_date": efinancial.get("posted"),
    "description": efinancial["job_description"],
    "rating": np.nan,
    "source": "efinancialcareers",
    "url": efinancial["url"]
})
```

```python
mycareers_clean = pd.DataFrame({
    "title": mycareers["title"],
    "company": mycareers["company"],
    "location": mycareers["location"],
    "employment_type": mycareers.get("employment_type"),
    "job_type": mycareers.get("job_level"),
    "salary": mycareers.get("salary"),
    "posted_date": mycareers.get("posted_date"),
    "description": mycareers["roles_and_responsibilities"],
    "rating": np.nan,
    "source": "mycareersfuture",
    "url": mycareers["url"]
})
```

```python
combined_jobs = pd.concat([
    glassdoor_clean,
    indeed_clean,
    prosple_clean,
    jobstreet_clean,
    efinancial_clean,
    mycareers_clean
], ignore_index=True)
```
